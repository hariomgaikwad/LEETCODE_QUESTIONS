Step 1: Node Class

Visual:

+-------------------+
| data = 10         |
| next = ?          |
+-------------------+

```python
class Node:

    def __init__(self, data):

        self.data = data 
        self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

# Data Access
print(node1.data)
print(node2.data)
print(node3.data)

# Next Access
print(node1.next.data)
print(node1.next.next.data)
```
