Implementing a DLL Node

```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None
        self.prev = None


node1 = Node(100)
node2 = Node(200)
node3 = Node(300)
node4 = Node(400)

# Forward Links
node1.next = node2
node2.next = node3
node3.next = node4

# Backward Links
node2.prev = node1
node3.prev = node2
node4.prev = node3

# Access Data
print(node1.data)
print(node2.data)
print(node3.data)
print(node1.next.data)
print(node1.next.next.data)
print("------------")

# Prev Pointer
print(node3.prev.data)
print(node3.prev.prev.data)
print(node1.next.data)
print(node2.prev.data)
```

Insert at Head 

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None 

class DoublyLinkedList:

    def __init__(self):
        self.head = None 

    # Insert at Head 
    def insert_head(self, data):

        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return
        
        new_node.next = self.head

        self.head.prev = new_node

        self.head = new_node

# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3

# dll.insert_head(30)
# dll.insert_head(20)
dll.insert_head(10)

# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
print(dll.head.next.next.next.data)
print(dll.head.next.next.next.prev.data)
```

Append (Insert at End)

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None 

class DoublyLinkedList:

    def __init__(self):
        self.head = None 

    # Append (Insert at End)
    def append(self, data):

        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return 
        
        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node
        new_node.prev = current 

# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3

dll.append(10)
# dll.append(20)
# dll.append(30)
# dll.append(40)

# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
print(dll.head.next.next.next.data)
print(dll.head.next.next.next.prev.data)
print("-------")
# For Access of Apeend 
print(dll.head.next.next.next.next.data)
print(dll.head.next.next.next.next.prev.data)
```

Insert at Position

```python
class Node:

    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:

    def __init__(self):
        self.head = None

    # Insert At Position 
    def insert_at_position(self, data, position):

        if position == 0:
            self.insert_head(data)
            return 
        
        new_node = Node(data)

        current = self.head
        count = 0

        while current is not None and count < position - 1:
            current = current.next
            count += 1
        
        if current is None:
            print("Invalid Position")
            return 
        
        prev_node = current.prev

        new_node.next = current
        new_node.prev = prev_node

        prev_node.next = new_node
        current.prev = new_node


# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3

# For call insert of position 
dll.insert_at_position(30,2)

# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
print(dll.head.next.next.next.data)
print(dll.head.next.next.next.prev.data)
print("-------")

dll.display_forward()
        
```

```python
class Node:

    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:

    def __init__(self):
        self.head = None

    # Forward Traverse
    def display_forward(self):

        current = self.head

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.next

        print("None")
# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3


# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
print(dll.head.next.next.next.data)
print(dll.head.next.next.next.prev.data)

# Call for forward 
dll.display_forward()
```

✅ Backward Traverse

```python
class Node:

    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:

    def __init__(self):
        self.head = None

    # Backward Traverse
    def display_backward(self):

        if self.head is None:
            print("List is Empty")
            return 
        
        current = self.head

        while current.next is not None:
            current = current.next

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.prev
        
        print("None")

# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3


# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
print(dll.head.next.next.next.data)
print(dll.head.next.next.next.prev.data)

# Call for Backward Traversal
dll.display_backward()
```

Delete Head (Delete First Node)

```python
class Node:

    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:

    def __init__(self):
        self.head = None

    # Forward 

    def display_forward(self):

        current  = self.head

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.next

        print("None")

    # Delete Head 

    def delete_head(self):

        if self.head is None:
            print("List is Empty")
            return
        
        if self.head.next is None:
            self.head = None
            return 
         
        self.head = self.head.next
        self.head.prev = None

# Object 
dll = DoublyLinkedList()

# Input of Nodes
node1 = Node(111)
node2 = Node(222)
node3 = Node(333)
node4 = Node(444)

# Joind A Node 
dll.head = node1

# Join node 1 and 2
node1.next = node2
node2.prev = node1

# Join node 2 and 3
node2.next = node3
node3.prev = node2

# Join node 3 and 4
node3.next = node4
node4.prev = node3

dll.delete_head()

# Access of Data 
print(dll.head.data)
print(dll.head.prev)
print("------")
print(dll.head.next.data)
print(dll.head.next.prev.data)
print("------")
print(dll.head.next.next.data)
print(dll.head.next.next.prev.data)
print("-------")
# print(dll.head.next.next.next.data)
# print(dll.head.next.next.next.prev.data)

# call for Forward
dll. display_forward()
```
