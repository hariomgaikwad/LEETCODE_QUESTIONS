You are given the head of a Singly Linked List and a value x, insert that value x at the end of the LinkedList and return the head of the modified Linked List.

Examples :

Input: x = 6,
   
Output: 1 -> 2 -> 3 -> 4 -> 5 -> 6
Explanation: We can see that 6 is inserted at the end of the linkedlist.
   
Input: x = 1,
   
Output: 4 -> 5 -> 1
Explanation: We can see that 1 is inserted at the end of the linked list.
      
Constraints:

0 ≤ size of linked list, node.data, x ≤ 105


```python
def Linked_List_End_Insertion(head, x):
    new_node = None(x)

    if head is None:
        return new_node

    current = head 
    while current.next is not None:
        current = current.next

    current.next = new_node

    return head 

# head = 10
x = 6

print(Linked_List_End_Insertion(head, x))
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


def Linked_List_End_Insertion(head, x):
    new_node = Node(x)

    if head is None:
        return new_node

    current = head

    while current.next is not None:
        current = current.next

    current.next = new_node

    return head


# Create the original linked list: 10 -> 20 -> None
head = Node(10)
head.next = Node(20)

x = 6

head = Linked_List_End_Insertion(head, x)

# Print the linked list
current = head
while current is not None:
    print(current.data, end=" -> ")
    current = current.next

print("None")

head = Node(10)
head.next = Node(20)

head = Linked_List_End_Insertion(head, 30)
```
