 BY USE FOR APPEND METHOD
NOTE : append for last

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    def append(self, data):

        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

ll = SinglyLinkedList()

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)


ll.head = node1
node1.next = node2
node2.next = node3

# Access of Data
print(ll.head.data)
print(ll.head.next.data)
print(ll.head.next.next.data)

ll.append(100)
ll.append(200)
ll.append(300)
ll.append(400)

print(ll.head.data)
print(ll.head.next.data)
print(ll.head.next.next.next.data)

```

Traversal 

```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None

class  Singly_Linked_List:

    def __init__(self):
        self.head = None

    # BY USE FOR APPEND
     
    def append(self, data):
        new_node = Node(data)
         
         # use for empty node using
        if self.head is None:
            self.head = new_node
            return 
        
        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    
    # BY USE FOR TRAVERSAL METHOD
    def display(self):

        if self.head is None:
            print("Linked List is Empty")
            return 
        current = self.head

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")

# Object 
A = Singly_Linked_List()

node1 = Node(20)
node2 = Node(40)
node3 = Node(60)
node4 = Node(80)

A.head = node1
node1.next = node2
node2.next = node3
node3.next = node4

# Access data
print(A.head.data)
print(A.head.next.data)
print(A.head.next.next.data)
print(A.head.next.next.next.data)


# Append of Node
A.append(100)

# Access of Traversal
A.display()
```

```python
class Node:

    def __init__(self, data):

        self.data = data
        self.next = None

class  Singly_Linked_List:

    def __init__(self):
        self.head = None

    def inser_at(self, val, position):
        new_node = Node(val)

        if position == 0:
            new_node.next = self.head
            self.head = new_node

        else:
            current = self.head
            prev_node = None
            count = 0

            while current is not None and count < position:
                prev_node = current
                current = current.next
                count += 1

            prev_node.next = new_node
            new_node.next = current 

# Object 
A = Singly_Linked_List()

node1 = Node(20)
node2 = Node(40)
node3 = Node(60)
node4 = Node(80)

A.head = node1
node1.next = node2
node2.next = node3
node3.next = node4
A.inser_at(50,2)

# Access data
print(A.head.data)
print(A.head.next.data)
print(A.head.next.next.data)
print(A.head.next.next.next.data)
```
