283. Move Zeroes

Given an integer array nums, move all 0's to the end of it while maintaining the relative order of the non-zero elements.

Note that you must do this in-place without making a copy of the array.

Example 1:
Input: nums = [0,1,0,3,12]
Output: [1,3,12,0,0]

Example 2:
Input: nums = [0]
Output: [0]

```python
# Brute Force Solution

def Move_Zero(nums):
    n = len(nums)
    temp = []

    for i in range(0,n):
        if nums[i] != 0:
            temp.append(nums[i])

    nz = len(temp)
    for i in range(0,nz):
        nums[i] = temp[i]

    for i in range(nz, n):
        nums[i] = 0

nums = [0, 1, 0, 3, 12]
print(Move_Zero(nums))
print(nums)

# TC = O(n)
# SC = O(n)
```

```python
# Optimal Approch 

def Move_Zero(nums):

    if len(nums) == 1:
        return 
    i = 0
    while i < len(nums):
        if nums[i] == 0:
            break
        i += 1

    if i == len(nums):
        return 
    j = i + 1

    while j <len(nums):
        if nums[j] != 0:
            nums[i],nums[j] = nums[j],nums[i]
            i += 1
        j += 1

nums = [0, 1, 0, 3, 12]
print(Move_Zero(nums))
print(nums)       
```
