2348. Number of Zero-Filled Subarrays

Given an integer array nums, return the number of subarrays filled with 0.

A subarray is a contiguous non-empty sequence of elements within an array.

Example 1:

Input: nums = [1,3,0,0,2,0,0,4]
Output: 6
Explanation: 
There are 4 occurrences of [0] as a subarray.
There are 2 occurrences of [0,0] as a subarray.
There is no occurrence of a subarray with a size more than 2 filled with 0. Therefore, we return 6.

```python
# 1️⃣ Brute Force
def Zeros_Subarrays(nums):
        total = 0
        n = len(nums)

        for i in range(n):
            for j in range(i, n):
                if nums[j] == 0:
                    total += 1
                else:
                    break

        return total

nums = [1, 3, 0, 0, 2, 0, 0, 4]
print("Count of SubArrays:",Zeros_Subarrays(nums))
print(nums)
```

```python
# 2️⃣ Formula Approach — k * (k + 1) // 2

def Zeros_Subarrays(nums):

    count = 0
    total = 0

    for num in nums:
        if num == 0:
            count += 1
        else:
            total += count * (count + 1) // 2
            count = 0

    # total += count * (count + 1) // 2

    return total

nums = [1, 3, 0, 0, 2, 0, 0, 4]
print("Count of SubArrays:",Zeros_Subarrays(nums))
print(nums)

# Time: O(n)
# Space: O(1)
```

```python
#3️⃣ Running Count Approach 🔥
def Zeros_Subarrays(nums):      
        count = 0
        total = 0

        for num in nums:
            if num == 0:
                count += 1
                total += count
            else:
                count = 0

        return total

nums = [1, 3, 0, 0, 2, 0, 0, 0]
print("Count of SubArrays:",Zeros_Subarrays(nums))
print(nums)
```
