27. Remove Element

Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.

Consider the number of elements in nums which are not equal to val be k, to get accepted, you need to do the following things:

Change the array nums such that the first k elements of nums contain the elements which are not equal to val. The remaining elements of nums are not important as well as the size of nums.
Return k.

Example 1:

Input: nums = [3,2,2,3], val = 3
Output: 2, nums = [2,2,_,_]
Explanation: Your function should return k = 2, with the first two elements of nums being 2.
It does not matter what you leave beyond the returned k (hence they are underscores).
Example 2:

Input: nums = [0,1,2,2,3,0,4,2], val = 2
Output: 5, nums = [0,1,4,0,3,_,_,_]
Explanation: Your function should return k = 5, with the first five elements of nums containing 0, 0, 1, 3, and 4.
Note that the five elements can be returned in any order.
It does not matter what you leave beyond the returned k (hence they are underscores).

```python
# Brute force solution logic

# Appn temp nava cha ek list ghetoy 
# aapn ek loop vaprtoy ka ?
# Jr nums of i not eual to value asl tr temp madhi aad kru
# jevdya kahi not equal chya values aahe tya temp madhi add hotil 

# Mg aapn prt ek loop vaprla jevdhi temp chi value aahe tya pryant 
# temp chya 1st element pasun suru kru 
# ani tyala nums mhanje array madhi add kru (origanal)
# aapn temp madhi jo element remove kraycha hota to kel ani baki add kele hote 
# mhanun last la temp chi len write kraychi aahe 
```

```python
def removeElement(nums, val):
    temp = []

    for num in nums:
        if num != val:
            temp.append(num)

    for i in range(len(temp)):
        nums[i] = temp[i]

    return len(temp)

nums = [3,2,2,1,4]
val = 3
print(removeElement(nums, val))
print(nums)
```

```python
# <!-- Better solution 
# step : 1 = k nava cha ek pointer ghetoy nums of i pasun 
# step : 2 = aapn ek loop chalu last pryant nums (n)
# step : 3 = condition ghetliy ki aapla jo i aahe nums of i (1st element of nums) 
#            jr to not equal asl valu (target) chay tr tyala k chya jagi takaycha 
#            mhanje ki swap kraycha 
# step : 4 = k la add kraych k += 1
# step : 5 = k mhanje aaplaya nums madhi values sodun jevdhe kahi elemetn aahe tyancha count  -->
```

```python
# Better Solution
def removeElement(nums, val):
    k = 0

    for i in range(len(nums)):

        if nums[i] != val:
            nums[k] = nums[i]
            k += 1

    return k

nums = [2,3,2,2,1]
val = 3
print(removeElement(nums, val))
print(nums)
```

```python
# Optimal Solution 

# step 1 : = (i = 0) ghetala ani nums chi len ghetliy 
# step 2 : = ek loop lavla applyala la target sodun baki kiti elemet aahe te sodhayche aahe 
# step 3 : = Condition dile ki jeva prynt i < n asl to pryant loop chalala loop true hou pryant 
# step 4 : = jr nums madhi 1st element value chay equal asl tr aapn last 
#             vala element 1st valay madhi takaycha ani to last la takaycha 
# step 5 : = aapni lenth aahe na nums chi tyala ek ni kami kraycha (n -= 1)
# step 6 : = else madhi i += 1 kraycha 
# step 7 : = lat la n return kraycha 
```

```python
# Optimal solution
def removeElement(nums, val):
    i = 0
    n = len(nums)

    while i < n:

        if nums[i] == val:
            nums[i] = nums[n - 1]
            n -= 1
        else:
            i += 1

    return n

nums = [3,2,2,1]
val = 3
print(removeElement(nums, val))
print(nums)
```
