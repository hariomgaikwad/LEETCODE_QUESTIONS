Largest in Array

Given an array arr[]. The task is to find the largest element and return it.

Examples:
Input: arr[] = [1, 8, 7, 56, 90]
Output: 90
Explanation: The largest element of the given array is 90.

```python
# Brute Solution

def max_number(nums):
    largest = nums[0]
    n = len(nums)

    for i in range(0, n):
        largest = max(largest, nums[i])
    return largest

nums = [55, 32, -97, 99, 3, 67]
print(max_number(nums))

# Time Complexity =  O(n)
# Space Complexity = O(1)
```

```python
def max_number(nums):
    largest = nums[0]
    n = len(nums)

    for i in range(0, n):
        if nums[i] > largest:
            largest = nums[i]

    return largest

nums = [55, 32, -97, 99, 3, 67]
print(max_number(nums))

# 👉 Time Complexity: O(n)
# 👉 Space Complexity: O(1)
```
