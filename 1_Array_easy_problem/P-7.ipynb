Rotate Array by One

Given an array arr, rotate the array by one position in clockwise direction.

Examples:
Input: arr[] = [1, 2, 3, 4, 5]
Output: [5, 1, 2, 3, 4]
Explanation: If we rotate arr by one position in clockwise 5 come to the front and remaining those are shifted to the end.
<!--  -->
Input: arr[] = [9, 8, 7, 6, 4, 2, 1, 3]
Output: [3, 9, 8, 7, 6, 4, 2, 1]
Explanation: After rotating clock-wise 3 comes in first position.

```python
def Rotate_Array(nums):
    n = len(nums)

    # nums[:] = [nums[-1]] + nums[0:n-1] # 1st method
    nums[:] = [nums[n-1]] + nums[0:n-1]  # 2nd method 

    # nums[:] म्हणजे original list मधले सगळे elements
    # नवीन values ने replace करणे (in-place update).


nums = [9, 8, 7, 6, 4, 2, 1, 3]

print(Rotate_Array(nums))
print(nums)

# TC = O(n)
# SC = O(1)
```

```python
# Better Solution 

def Rotate_Array(nums):
    n = len(nums)
    temp = nums[n-1]

    for i in range(n-2,-1,-1):
        nums[i+1] = nums[i]
        nums[0] = temp


nums = [9, 8, 7, 6, 4, 2, 1, 3]

print(Rotate_Array(nums))
print(nums)

# TC = O(n)
# SC = O(1)
```
